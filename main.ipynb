```python
from ultralytics import YOLO
```

```python
import torch
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
path = "boneFractureDetection-best.pt"
```

```python
image = "assets/7_mri_0_2623_jpeg.rf.c9228500a8267cbf652681ef66377f33.jpg"
```

```python
model = YOLO(model=path)
```

```python
results = model(image, conf=0.25)[0]
```

```python
import cv2
```

```python
img = cv2.imread(image)
```

```python
for box in results.boxes:
    conf = float(box.conf[0])
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    cls = int(box.cls)
    label = model.names[cls]
    
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
    cv2.putText(img, f"{label} {conf:.2f}", (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

cv2.imshow("Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
